```python
###Cambiar tipo de dato(str) a numérico, en este caso(float)

Base_de_datos['Global_Sales'] = Base_de_datos['Global_Sales'].str.replace(',','.')
Base_de_datos['Global_Sales'] = Base_de_datos['Global_Sales'].astype(float) 
```

```python
Base_de_datos['EU_Sales'] = Base_de_datos['EU_Sales'].str.replace(',','.')
Base_de_datos['EU_Sales'] = Base_de_datos['EU_Sales'].astype(float)
```

```python
Base_de_datos['NA_Sales'] = Base_de_datos['NA_Sales'].str.replace(',','.')
Base_de_datos['NA_Sales'] = Base_de_datos['NA_Sales'].astype(float)
```

```python
Base_de_datos['JP_Sales'] = Base_de_datos['JP_Sales'].str.replace(',','.')
Base_de_datos['JP_Sales'] = Base_de_datos['JP_Sales'].astype(float)
```

```python
Base_de_datos['Other_Sales'] = Base_de_datos['Other_Sales'].str.replace(',','.')
Base_de_datos['Other_Sales'] = Base_de_datos['Other_Sales'].astype(float)
```

```python
Base_de_datos[['Global_Sales','EU_Sales','NA_Sales','JP_Sales','Other_Sales']].sum() 
```

```python
#PRIMERA PREGUNTA 1.¿Cuál es la plataforma que tiene más ventas a nivel global?

#Use groupby para agrupar las dos columnas necesarias para responder a mi pregunta. .sum() para sumar todas las ventas por plataforma
#finalmente .idmax que busca el valor más alto y me indica a qué plataforma pertenece.

print ('La plataforma más vendida  a nivel global es:')
print (Base_de_datos.groupby('Platform')['Global_Sales'].sum().idxmax())
'-------------------------------------------------------------------------'
```

```python
#SEGUNDA PREGUNTA 2.¿Cuáles son los primeros 10 Publisher que tienen la mejor clasificación de puntos según la critica?
#Primero vamos a definir el orden de importancia creando un diccionario.

Order_points = {
    "Excelente": 4,
    "Bueno": 3,
    "Aceptable": 2,
    "Malo": 1
}
```

```python

#Agregaremos una columna nueva llamada Score_Num, selecionaremos de nuevo nuestra variable, insertamos la columna
#Utilizaremos .map, lo cual nos sirve como traductor de cada valor de la columna seleccionada previamente, gracias al diccionario creado en la celda anterior(Order_points)

Base_de_datos["Score_Num"] = Base_de_datos["Critic_Score_Class"].map(Order_points)
```

```python
 
#Ahora agruparemos con groupby la columna de publisher y la nueva columna creada de puntuación numérica, 
# usamos.mean() para calcular el promedio para cada publisher
#.sort_values(ascending=FaLse) Ordena esos promedios de mayor a menor
#.Head(11) Muestra los primeros 10 resultados.

print('Estos son los 10 Publisher con mejor clasificación de puntos según la critica:')
Base_de_datos.groupby('Publisher')['Score_Num'].mean().sort_values(ascending=False).head(10)


```

```python
#TECERA PREGUNTA 3. ¿Cuál es el género con mayor rentabilidad en Norte América y Japón?
#Aquì usé groupby para agrupar las columnas respectivas /genre, Na_Sales y JP_Sales, .Sum para sumar por género y .sum(axis=1)para sumar 
#los valores de las filas (NA Y JP) para cada género.
#Finalmente .idxmax para identificar cuál de todos los generos en mi base de datos entre ambas columnas tiene la cifra màs alta

print('El género con mayor rentabilidad en Norte América y Japón es:')

Base_de_datos.groupby('Genre')[['NA_Sales', 'JP_Sales']].sum ().sum(axis=1).idxmax()



```

```python
###CONCLUSIÓN (No use codificación muy avanzada ya que considero que quería basarme en algo que yo entendiera e ir subiendo mi nivel pero siempre entendiendo lo que estoy haciendo.
print('CONCLUSIÓN: En el Dataset podemos evidenciar las siguientes afirmaciones:' \
' 1.La plataforma con más ventas a nivel global es la Play Station 2' \
'2. Según la clasificación por critica, el publisher con mejor puntaje es InXile Intertainment, con puntaje excelente' \
'3. El género con mayor rentabilidad en Norte América y Japón es Acción')
```
