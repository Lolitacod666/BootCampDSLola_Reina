```python
import requests
```

```python
resp = requests.get("https://api-colombia.com/api/v1/NaturalArea") 
```

```python
resp.json()
```

```python
import pandas as pd

df = pd.DataFrame(resp.json())
```

```python
df = df.drop(['department'], axis=1)

df = df.drop(['categoryNaturalArea'], axis=1)
```

```python
df['landArea'] = df['landArea'].fillna('No Asignado')

df['daneCode'] = df['daneCode'].fillna(0)

df['departmentId'] = df['departmentId'].fillna(0)

```

```python
df['maritimeArea'] = df['maritimeArea'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df['landArea']= pd.to_numeric(df['landArea'], errors='coerce')

df.sort_values(by='landArea', ascending=False).head(5)
```

```python
df['landArea'] = pd.to_numeric(df['landArea'], errors='coerce').fillna(0)
pd.set_option('display.float_format', lambda x: '%.2f' % x)
df['Hectareas_Terrestres'] = df['landArea'] / 10000
 
```
